## Яркостные преобразования

### Линейное преобразование

Кратко: линейное преобразование изображения – это преобразование вида: $$y = \alpha x + \beta, \ \forall x \in I,$$ где $I$ – набор пикселей изображения.

````python
# TODO: Реализуйте функцию для линейного преобразования. Примените его к изображению.
# TODO: Поэкспериментируйте с коэффициентами альфа и бета (альфа >, < 1, бета >, < 0)

def linear_transformation(image_path=image, alpha=1.0, beta=0.0):
    image_def = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    image_float = image_def.astype(np.float32)
    transformed = alpha * image_float + beta
    transformed = np.clip(transformed, 0, 255)

    cv2_imshow(transformed)


input_image = input('Путь к изображению (Enter=загруженный мной)')

for i in range(5):
  random_contrast = randint(0, 10)
  random_brightness = randint(0, 100)
  print('\n' + ' ' * 20 + '=' * 60)
  print(' ' * 37 + f'КОНТРАСТ = {random_contrast}, ЯРКОСТЬ = {random_brightness}')
  print(' ' * 20 + '=' * 60)
  print('\n' * 2)
  linear_transformation(input_image, random_contrast, random_brightness)
````

### Гамма-коррекция

Кратко: гамма коррекция – это преобразование вида: $$y = Cx^{\gamma},\ \forall x \in I$$

````python
# TODO: Реализуйте функцию для гамма-коррекции. Примените его к изображению.
# TODO: Поэкспериментируйте с коэффициентами C и гамма (C >, < 1, гамма >, < 1)

def gamma_correction(image_path=image, gamma=1.0, c=1.0):
    image_def = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    image_normalized = image_def.astype(np.float32) / 255.0
    corrected = c * image_normalized ** gamma
    corrected = np.clip(corrected, 0, 1) * 255

    cv2_imshow(corrected)


input_image = input('Путь к изображению (Enter=загруженный мной)')

for i in range(5):
  random_gamma = randint(1, 12)
  random_scale_coeff = randint(1, 12)
  print('\n' + ' ' * 20 + '=' * 60)
  print(' ' * 15 + f'ГАММА = {random_gamma}, КОЭФИЦИЕНТ МАШТАБИРОВАНИЯ = {random_scale_coeff}')
  print(' ' * 20 + '=' * 60)
  print('\n' * 2)
  gamma_correction(input_image, random_gamma, random_scale_coeff)
````

## Серый мир

Говорят, при смешении всех цветов в мире должен получиться серый... :(

Не волнуйтесь, это не правда, средний цвет вселенной – [космический латте](https://ru.wikipedia.org/wiki/%D0%9A%D0%BE%D1%81%D0%BC%D0%B8%D1%87%D0%B5%D1%81%D0%BA%D0%B8%D0%B9_%D0%BB%D0%B0%D1%82%D1%82%D0%B5).

Тем не менее, существует преобразование "серый мир". Суть – взять среднее каждого канала, найти их общее среднее, найти коэффициент масштабирования для каждого канала (общее среднее поделить на среднее канала) и умножить каждую компоненту пикселя на соответствующий коэффициент:
$$m = \frac {R_{mean} + G_{mean} + B_{mean}} {3} \\
  S_R = \frac {m} {R_{mean}} \\
  S_G = \frac {m} {G_{mean}} \\
  S_B = \frac {m} {B_{mean}} \\
  y_R = x_R \cdot S_R, \ \forall x_R \in I_R \\
  y_G = x_G \cdot S_G, \ \forall x_G \in I_G \\
  y_B = x_B \cdot S_B, \ \forall x_B \in I_B,$$ где $I_R, I_G, I_B$ – соответствующие каналы изображения $I$

## Размытия изображений

Важнейшей частью обработки изображений являются разнообразные фильтры, выполняемые с помощью [свёрток](https://ru.wikipedia.org/wiki/%D0%A1%D0%B2%D1%91%D1%80%D1%82%D0%BA%D0%B0_(%D0%BC%D0%B0%D1%82%D0%B5%D0%BC%D0%B0%D1%82%D0%B8%D1%87%D0%B5%D1%81%D0%BA%D0%B8%D0%B9_%D0%B0%D0%BD%D0%B0%D0%BB%D0%B8%D0%B7). Не углубляйтесь в детали, просто представляйте себе, что свёртка - это когда одна картинка "окном" ходит по другой и делает линейную комбинацию "наложением".

Самыми базовыми являются размытия. Они используются, чтобы убрать шумы в изображении (но не только).

Мы сегодня рассмотрим фильтр по Гауссу и медианный. Фильтр по Гауссу в качестве "окна" (оно зовётся "ядром" свёртки) использует [двумерное нормальное распределение](https://ru.wikipedia.org/wiki/%D0%9C%D0%BD%D0%BE%D0%B3%D0%BE%D0%BC%D0%B5%D1%80%D0%BD%D0%BE%D0%B5_%D0%BD%D0%BE%D1%80%D0%BC%D0%B0%D0%BB%D1%8C%D0%BD%D0%BE%D0%B5_%D1%80%D0%B0%D1%81%D0%BF%D1%80%D0%B5%D0%B4%D0%B5%D0%BB%D0%B5%D0%BD%D0%B8%D0%B5). Звучит страшно, но на деле это выглядит как [яркая лампочка](https://www.researchgate.net/publication/320108701/figure/fig1/AS:11431281335665633@1743373673366/Numerical-simulation-sample-Gaussian-blur-kernels-used-to-model-projector-defocusing.jpg).

Медианный фильтр, вообще говоря, не свёртка. Он просто ходит "окном" по картинке и в каждый момент выбирает медианное значение среди пикселей, попавших в "окно".


Медианный фильтр иногда оставляет более резкое и чёткое изображение, но у него плохие свойства, в отличие от размытия по Гауссу
